```python
# spark
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName("Shopping Trends") \
    .getOrCreate()
```

```python
# load data
!wget -q https://www.andrew.cmu.edu/user/mfarag/static/shopping_trends.csv -O shopping_trends.csv
```

```python
# use spark to read the data
df = spark.read.csv(
    "shopping_trends.csv",
    header=True,
    inferSchema=True
)
df.show(5)
df.printSchema()
```
